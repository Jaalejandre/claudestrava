```python
# Module 1: Physical Constants and Lennard-Jones Potential
import numpy as np
import matplotlib.pyplot as plt

def lennard_jones(r, sigma=0.3166, eps=0.6502):
    sr6 = (sigma / r)**6
    return 4.0 * eps * (sr6**2 - sr6)

r = np.linspace(0.28, 0.80, 200)
u = lennard_jones(r)

plt.figure(figsize=(8,4))
plt.plot(r, u, 'b-', lw=2, label='LJ Potential (SPC/E Water Oxygen)')
plt.axhline(0, color='gray', ls='--')
plt.xlabel('Distance r (nm)')
plt.ylabel('Potential Energy U (kJ/mol)')
plt.title('Lennard-Jones Interatomic Potential')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```
