```python
# Module 2: Surface Tension Calculation from Pressure Tensor
import numpy as np

def compute_surface_tension(p_normal, p_tangential, box_z_nm=10.5):
    # gamma = 0.5 * Lz * (PN - PT) * 0.1 mN/m
    gamma = 0.5 * box_z_nm * (p_normal - p_tangential) * 0.1
    return gamma

P_zz = -50.0 # bar (Normal)
P_T = -520.0  # bar (Tangential in liquid film)
gamma = compute_surface_tension(P_zz, P_T)
print(f'Calculated Liquid-Vapor Surface Tension: {gamma:.2f} mN/m')
```
